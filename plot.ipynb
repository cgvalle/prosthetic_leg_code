```python
from matplotlib import pyplot as plt
import pandas as pd
```

```python
angle = pd.read_csv('knee_angle_gait.csv')


plt.plot(angle['cycle'], angle['angle'])
```

```python
angles = pd.read_csv('angles.txt')

plt.plot(angles['time'], angles['target_angle'], label='target')
plt.plot(angles['time'], angles['current_angle'], label='current')
plt.xlabel('time (s)')
plt.ylabel('angle (deg)')
plt.legend()
plt.show()
```
